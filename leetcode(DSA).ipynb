#### 1652. Defuse the Bomb

```python
code = [5,7,1,4]
k = 3
n=len(code)
res=[0]*n

if k==0:
    print(res)
elif k>0:
    window_sum=sum(code[1:k+1])
    
    for i in range(n):
        res[i]=window_sum
        window_sum-=code[(i+1)%n]
        window_sum+=code[(i+1+k)%n]
    print(res)


else:
    k=-k
    window_sum=sum(code[k:n+1])

    for i in range(n):
        res[i]=window_sum

        window_sum-=code[(i-k)%n]
        window_sum+=code[i]

    print(res)
```
