```python
from datetime import datetime
from zoneinfo import ZoneInfo

# Get the time in a specific timezone, e.g., 'Asia/Kolkata'
tz_india = ZoneInfo('Asia/Kolkata')
c = datetime.now(tz_india)
c=int(c.strftime("%H"))
print(type(c))
```
