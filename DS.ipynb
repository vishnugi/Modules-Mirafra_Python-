### implimentation of stack:

```python
stack=[]
while True:
    print("1.PUSH 2.POP 3.PEEK 4.EXIT")
    ch=int(input())
    match ch:
        case 1:
            stack.append(int(input()))

        case 2:
            if len(stack)==0:
                print("Stack in empty")
            else:
                print(stack.pop())
        case 3:
            if len(stack)==0:
                print("Empty stack")
            else:
                print(stack[-1])
        case _ :
            break
print(stack)
```

#### Balancing paranthesis 

```python
s="{()}[]"
s=list(s)
stack=[]
for ch in s:
    if ch in "{[(":
        stack.append(ch)
    else:
        if len(stack)==0:
            print("yes")
            break
        else:
            if ch==")" and stack[-1]=="(":
                stack.pop()
            elif ch=="]" and stack[-1]=="[":
                stack.pop()
            elif ch=="}" and stack[-1]=="{":
                stack.pop()
            else:
                print("No")
                break
if len(stack)==0:
    print('yes')
else:
    print("NO")
```
