## Log File Automation 

##### Extracting only errors 

```python
def check_error(file_path):
    file= open('sensor.log','r')
    count={'ERROR':0,'INFO':0,'WARNING':0}
    for line in file:
        if 'ERROR' in line:
            count['ERROR']+=1
    file.close()
    return count
```

```python
ans=check_error('sensor.log')
print(ans)
```

##### Extracting timestamps and messages :

```python
def timestamp_message(file_path):
    with open('sensor.log') as file:
        for line in file :
            part= line.split(" ",2)
            # print("part:",part)
            timestamp=part[0]+" "+part[1]
            message= part[2]

            print("Message:",message)
            print("Timestamp:",timestamp)
```
