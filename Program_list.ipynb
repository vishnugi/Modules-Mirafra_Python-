## List

```python
# Input: [1,2,3,4,5] → Output: Sum of all elements
# Expected Output: 15

l=list(map(int,input().split()))
ans=0
for i in range(len(l)):
    ans=ans+l[i]
print(ans)
    
```

```python
# Input: [10,20,30,40,50], n=3 → Output: First n elements
# Expected Output: [10,20,30]

l=list(map(int,input().split()))
n=int(input())
ans=[]
for i in range(n):
    ans.append(l[i])
print(ans)
```

```python
# Input: [3,1,4,1,5,9,2,6] → Output: Remove duplicates, keep order
# Expected Output: [3,1,4,5,9,2,6]

l=list(map(int,input().split()))
res=[]
for i in l:
    if i not in res:
        res.append(i)
print(res)
```

```python
# Input: [[1,2],[3,4],[5,6]] → Output: Flatten to single list
# Expected Output: [1,2,3,4,5,6]

l=[[1,2],[3,4],[5,6]]
ans=[]
for row in l:
    for i in row:
        ans.append(i)
print(ans)
```

```python
# Input: [1,2,3,4,5,6] → Output: Rotate right by 2
# Expected Output: [5,6,1,2,3,4]

l=list(map(int,input().split()))
k=int(input())
ans=l[len(l)-k::]+l[:len(l)-k:]
print(ans)
```

```python
# Input: [1,2,3,2,1,4,5,4] → Output: Find first non-repeating element
# Expected Output: 3

l=list(map(int,input().split()))
for i in l:
    if l.count(i)==1:
        print(i)
        break
```

```python
# Input: [1,2,3,4,5,6,7], k=3 → Output: Max sum of any subarray of size k
# Expected Output: 18

l=list(map(int,input().split()))
k=int(input())
res=0
for i in range(len(l)-k+1):
    ans = sum(l[i:k+i])
    print(l[i:k+i], ans)
    if res<ans:
        res=ans
print(res)
    
```

```python
# Input: [1,0,2,0,3,0,4] → Output: Move zeros to end without order change
# Expected Output: [1,2,3,4,0,0,0]

l=list(map(int,input().split()))
for i in range(len(l)):
    if l[i]==0:
        l.append(l.pop(i))
        
print(l)
```

```python
# Input: [(1,2),(3,4),(5,6)] → Output: Unzip into two tuples
# Expected Output: (1,3,5), (2,4,6)

t=[(1,2),(3,4),(5,6)]
t1=()
t2=()
for row in t:
    t1=t1+(row[0],)
    t2=t2+(row[1],)
print(t1,t2)
```

```python
# Input: ( (1,2), (3,4), (5,6) ) → Output: Flatten (nested tuple)
# Expected Output: (1,2,3,4,5,6)
t=((1,2),(3,4),(5,6))
ans=()
for i in t:
    for j in i:
        ans=ans+(j,)
print(ans)
```

```python
# Input: {'a':1,'b':2,'c':3,'d':2} → Output: Find keys with duplicate values
# Expected Output: {'b','d'}

d={'a':1,'b':2,'c':3,'d':2}
dup=set()
org=set()
for v in d.values():
    if v in org :
        dup.add(v)
    else:
        org.add(v)

ans={k for k,v in d.items() if v in dup}
print(ans)
```
