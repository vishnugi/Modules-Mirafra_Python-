```python
#Check whether a string is a palindrome.
s="madame"
i=0
j=len(s)-1
while i<j:
    if s[i]==s[j]:
        i=i+1
        j=j-1
    else:
        print("No")
        break
else:
    print("Yes")
```

```python
#Check whether a number is a palindrome.
n=1210
num=n
rev=0
while n!=0:
    rem=n%10
    rev=(rev*10)+rem
    n=n//10
if rev==num:
    print("yes!!")
else:
    print("no")
```

```python
# Remove duplicates from a list.
l=[4,2,4,2,1,3,4,5,6,7,3,2,4,5]
seen=set()
for i in l:
    if i not in seen:
        seen.add(i)
print(seen)
```

```python
# Find duplicate elements in a list.
l=[4,2,4,2,1,3,4,5,6,7,3,2,4,5]
seen=set()
dup=set()
for i in l:
    if i in seen:
        dup.add(i)
    seen.add(i)
print(dup)
```
