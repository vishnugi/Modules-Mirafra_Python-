#### Timing Decorator

```python
import time

def timer(func):
    def wrapper(*args,**kargs):
        start=time.time()
        res=func(*args,**kargs)
        end=time.time()
        print(f"TIME: {end-start}")
        return res
    return wrapper

@timer
def compute():
    for _ in range(10000):
        pass

compute()
```

```python
import time
from functools import wraps  # 1. Import the fix

def timer(func):
    @wraps(func)             # 2. Add this right above the wrapper
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Time taken: {end - start}")
        return result
    return wrapper

@timer
def compute():
    """This function runs a heavy loop."""
    for _ in range(1000000):
        pass

# Now the metadata is preserved perfectly!
print(compute.__name__)  # Output: compute  ✅
print(compute.__doc__)   # Output: This function runs a heavy loop
```
