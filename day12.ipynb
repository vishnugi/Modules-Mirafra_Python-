Encapsulation:


```python
class Gmail:
    def __init__(self):
        self.username="vish1"
        self.__password="vish"
        
    def __authentication(self,username,password):
        if self.username==username and self.__password==password:
            return True
        return False
        
    def login(self,username, password):
        if self.__authentication(username,password):
            return f"Logged in successfully by {username}"
        return f"Failed to login by {username}"
        
```

```python
g=Gmail()
g.login("vish1","vish")
```

```python
class Myrange():
    def __init__(self,value):
        self.value=value
        self.cur=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.cur<self.value:
            self.cur+=1
            return self.cur*3
        else:
            raise StopIteration
            
```

```python
a=Myrange(3)
print(a)
```
