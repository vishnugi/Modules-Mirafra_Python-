```python
class Student:
    def __init__(self,name,dept):
        self.name=name
        self.dept=dept
    def disp(self):
        print(self.name,self.dept)
    def get_Name(self):
        return self.name
    def set_Name(self,name):
        self.name=name

s1=Student("Vishnu","CSE")
s2=Student("Priya","ECE")

s1.disp()
s1.set_Name("Priya")
s1.disp()
s2.get_Name()
```

```python
class Sample:
    c=30
    def __init__(self,x,y):
        self.a=x # a and b are the instance variable
        self.b=y
    def disp(s):
        print(s.a,s.b)

    @classmethod
    def cmethod(cls):
        print(cls.c)

    @staticmethod 
    def smethod():
        print(Sample.c)
        
obj1=Sample(10,20)
obj1.disp()
obj1.cmethod()
obj1.smethod()
```

#### count how many obj are created for a given class

```python
class Sample:
    c=0
    def __init__(s):
        Sample.c=Sample.c+1
    @staticmethod
    def getcount():
        return Sample.c
s1=Sample()
s2=Sample()
s3=s1
s4=Sample()

print(Sample.getcount())
```

```python
class Vehicle:
    def start(self):
        print("Vehicle is started")
    def stop(self):
        print("Vehicle is stoped")

class Car(Vehicle):
    def run(self):
        print("Car is running....")
C1=Car()
C1.start()
C1.run()
C1.stop()
```

```python
class A:
    def __init__(self,a,b):
        self.a=a
        self.b=b

    def disp(self):
        print(self.a,self.b,end=" ")

class B(A):
    def __init__(self,a,b,c,d):
        super().__init__(a,b)
        self.c=c
        self.d=d
    def show(self):
        super().disp()
        print(self.c,self.d)

obj1=B(1,2,3,4)
obj1.show()
```

```python
class Sample:
    def disp(self):
        print("class Sample")
class Simple(Sample):
    def show(self):
        # Sample.disp(self)
        super().disp()
        print("class Simple")

obj=Simple()
obj.show()
```

```python
from abc import *
class Vehicle(ABC):
    @abstractmethod 
    def start(s):
        pass
    @abstractmethod
    def stop(s):
        pass

class Car(Vehicle):
    def start(s):
        print("car started")
    def stop(s):
        print("car stoped")

obj=Car()
obj.start()
obj.stop()
```
