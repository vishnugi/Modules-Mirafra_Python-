```python
# 3. Count how many numbers from 1 to N are multiples of 3 or 5.
n=int(input())
count=0
for i in range(1,n+1):
    if i%3==0 or i%5==0:
        count+=1
print(count)
    
```

```python
# 4. Given N, print the sum of squares 1²+2²+…+N².
n=int(input())
s=0
for i in range(1,n+1):
    s=(i*i)+s
print(s)
    
```

```python
# 5. Given N, print the factorial using a loop (no math library).
n=int(input())
fct=1
for i in range(1,n+1):
    fct=i*fct
print(fct)
```

```python
# 6. Given a list of integers, count how many are positive, negative, and zero.
l=list(map(int,input().split()))
pos=0
neg=0
zero=0
for i in l:
    if i<0:
        neg+=1
    elif i>0:
        pos+=1
    else:
        zero+=1
print("negetive: ",neg)
print("positive: ",pos)
print("zero: ",zero)
```

```python
# 11. Given a list, print elements at even indices, then at odd indices.
l=list(map(int,input().split()))
even=[]
odd=[]
for i in range(len(l)):
    if i%2==0:
        even.append(l[i])
    else:
        odd.append(l[i])
print(even+odd)
        
```

```python
# 12. Given a list, print elements in reverse without using reversed() or slicing.
l=list(map(int,input().split()))
i=0
j=len(l)-1
for i in range(len(l)//2):
    l[i],l[j]=l[j],l[i]
    j=j-1
print(l)
```
