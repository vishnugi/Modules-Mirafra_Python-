#### 345. Reverse Vowels of a String

```python
s = "leetcode"
s = list(s)
i = 0
j = len(s)-1
while i<=j:
    if s[i] in 'aeiouAEIOU' and s[j] in 'aeiouAEIOU':
        s[i],s[j]=s[j],s[i]
        i=i+1
        j=j-1
    elif s[i] in 'aeiouAEIOU':
        j=j-1
    elif s[j] in 'aeiouAEIOU':
        i=i+1
    else:
        i=i+1
        j=j-1
print("".join(s))
        
```

#### 443. String Compression

```python
chars = ["a","b","b","c","c","c"]
c=0
res=''
for i in range(len(chars)-1):
    if chars[i]==chars[i+1]:
        c+=1
    else:
        if c==0:
            res=res+chars[i]
        else:
            res=res+chars[i]+str(c+1)
        c=0
res=res+chars[len(chars)-1]+str(c+1)
print(len(res))
```

#### 917. Reverse Only Letters

```python
s = "a-bC-dEf-ghIj"
s=list(s)
i=0
j=len(s)-1
while i<=j:
    if s[i]=="-":
        i=i+1
    elif s[j]=="-":
        j=j-1
    else:
        s[i],s[j]=s[j],s[i]
        i=i+1
        j=j-1
print("".join(s))
```

### 11. Container With Most Water

```python
height = [1,8,6,2,5,4,8,3,7]
i=0
j=len(height)-1
ans=[]
for i in range(len(height)):
    h=min(height[i],height[j])
    vol=h*(j-i)
    ans.append(vol)
print(max(ans))
```
