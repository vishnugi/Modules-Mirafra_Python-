```python
import threading
import time

def background_task():
    print("Background task started...")
    time.sleep(10)  # Simulating a 2-second download
    print("Background task finished!")

# Create the thread
thread = threading.Thread(target=background_task)

# 1. Start the thread
thread.start() 
print("This line prints immediately because start() does not block!")

# 2. Join the thread
thread.join()  
print("This line will ONLY print after the 2-second background task is totally done.")
```

```python
# Create thread that prints numbers 1-5
from threading import Thread

def fun():
    for i in range(1,6):
        print(i)

t1=Thread(target=fun)

t1.start()
t1.join()
```

```python
# Two threads printing alternately
from threading import Thread, Condition

condition=Condition()
turn=0

def thread1():
    global turn
    for i in range(1,10,2):
        with condition:
            while turn!=0:
                condition.wait()
            print("thread1:",i)
            turn=1
            condition.notify()

def thread2():
    global turn
    for i in range(2,11,2):
        with condition:
            while turn!=1:
                condition.wait()
            print("thread2:",i)
            turn=0
            condition.notify()

t1=Thread(target=thread1)
t2=Thread(target=thread2)

t1.start()
t2.start()
t1.join()
t2.join()
    
```
