```python
#1. Check if a string is a palindrome

s=input()
# print(s==s[::-1])
i=0
j=len(s)-1
while i<j:
    if s[i]!=s[j]:
        print("No")
        break
    else:
        i=i+1
        j=j-1
else:
    print("Yes")
```

```python
# K length palindrome substring:
s="bbabbda"
k=int(input())
for i in range(k):
    if s[i:i+k]==s[i:i+k][::-1]:
        print(s[i:i+k])
        break
    
```

```python
#longest palindrome substring 
s="cbbd"
k=len(s)
for i in range(k):
    if s[i:k-i]==s[i:k-i][::-1]:
        print(s[i:k-i])
        break
```

```python
s="abcmadamdefg"
k=len(s)
num=k
while k!=0:
    for i in range(num):
        for j in range(num):
            if s[i:num-j]==s[i:num-j][::-1]:
                print(s[i:num-j])
                break    
    k=k-1
```
