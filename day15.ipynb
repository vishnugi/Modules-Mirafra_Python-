### cProfile:

```python
import cProfile

def is_Prime(n):
    if n<=1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return True
    return True

def check():
    res=0
    for i in range(1000000):
        if is_Prime(i):
            res+=1
if __name__=="__main__":
    cProfile.run("check()",sort='time')
```
