```python
def print_stream(stream,n):
    if stream=="odd":
        c=1
        for i in range(c,n+1):
            if c%2!=0:
                print(c)
                c+=2
    else:
        c=0
        for i in range(c,n):
            if c%2==0:
                print(c)
                c+=2

n=int(input())
for i in range(n):
    stream,n=input().split(); n=int(n)
    (print_stream(stream,n))
```

```python
def validate(n):
    num=str(n)
    if len(num)==10:
        if (num[0]==7 or num[0]==8 or num[0]==9):
            return "YES"
        else:
            print(len(num),num[0])
            return "NO"
    else:
        print(len(num))
        return "NO"
n=int(input())
for i in range(n):
    num=int(input())
    print(validate(num))
```

```python
s=input()
lower_char=[]
upper_char=[]
even_num=[]
odd_num=[]
for i in s:
    if i.isdigit():
        if int(i)%2==0:
            even_num.append(i)
        else:
            odd_num.append(i)
    else:
        if i.isupper():
            upper_char.append(i)
        else:
            lower_char.append(i)
print(sorted(upper_char))
print(sorted(lower_char))
print(sorted(odd_num))
print(sorted(even_num))
```

#### Nested List

```python
from collections import defaultdict
record=defaultdict(list)
n=int(input())
for i in range(n):
    name=input()
    score=float(input())
    record[score].append(name)
print(record)
score=sorted(record)
ans=sorted(record[score[1]])
print(ans,sep='\n')
```

```python
s="BANANA"
stuart=[]
kevin=[]
for i in range(len(s)):
    for j in range(i,len(s)):
        if s[i] in "AEIOU":
            kevin.append(s[j:len(s)])
        else:
            stuart.append(s[j:len(s)])
if len(stuart)>len(kevin):
    print("Stuart",len(stuart))
else:
    print("Kevin",len(kevin))
```

```python
l=[1,2,3,4,5,6,7]
k=9
low=0
upper=len(l)-1
while low<upper:
    mid=(low+upper)//2
    if l[mid]==k:
        print("found")
        break
    elif l[mid]<k:
        low=mid+1
    else:
        upper=mid-1
else:
    print("no")
```
