```python
# 47. Check whether a number is strong number (sum of factorial of digits).

n=int(input())
num=n
s=0
while n!=0:
    rem=n%10
    ans=1
    while rem!=0:
        ans*=rem
        rem-=1
    s+=ans
    n=n//10

print('Strong' if num==s else'Not a Strong num')
        
```

```python
# 48. Print strong numbers up to N.
a=int(input())
for n in range(a+1):
    num=n
    s=0
    while n!=0:
        rem=n%10
        ans=1
        while rem!=0:
            ans*=rem
            rem-=1
        s+=ans
        n=n//10
    if s==num:
        print(num)
    
```
