### Pillars of OOP'S :

```python
# method overloading is not suported durectly
# it will support indirectly

class Sample:
    def sum(self,a,b,c):
        return a+b+c
    def sum(self,a,b):
        return a+b
```

```python
s=Sample()
s.sum(10,20,20) #as overloading will not support completely
```

```python
class Sample:
    def sum(self,a,b,c=0):
        return a+b+c #with the default argument it can achieve the overloading
```

```python
s=Sample()
s.sum(20,30,40)
```
