```python
# Even Numbers

def even(n):
    for i in range(n):
        if i%2==0:
            yield(i)
print(list(even(10)))
```

```python
# generate square of 5 

def square(n):
    for i in range(n):
        yield i**2

print(list(square(5)))
```

```python
# fibinocci of first 10 numbers:

def fib(n):
    i=1
    a,b=-1,1
    while i<=n:
        res=a+b
        yield res
        a,b=b,res
        i+=1
print(list(fib(5)))
```

```python
# prime number genertor:

def prime(n):
    for i in range(2,n):
        for j in range(2,i):
            if i!=j and i%j==0:
                break
        else:
            yield i
print(list(prime(10)))
        
            
```

```python
def gen():
    x = yield
    print(x)

g = gen()
next(g)        # start generator
g.send(5)     # prints 10
```
